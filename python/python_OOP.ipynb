# 1. 절차 지향 vs 객체지향

```python
# 절차지향 방식: 데이터(딕셔너리)와 함수가 분리됨
sensor = {'id': 'S1', 'temp': 24.5}

def show(sensor):
    print(f"{sensor['id']}: {sensor['temp']}")

show(sensor)


# 객체지향 방식: 데이터와 동작이 하나로 묶임
class Sensor:
    def __init__(self, id, temp):
        self.id = id; self.temp = temp
    def show(self):
        print(f'{self.id}: {self.temp}')

Sensor('S1', 24.5).show()
```

# 2. 클래스와 객체

```python
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id # 속성(attribute)
        self.value = value
    def show(self): # 메서드(method)
        print(f'{self.sensor_id}: {self.value}')

s1 = Sensor('S-01', 24.5)
s2 = Sensor('S-02', 30.1)
s1.show() # S-01: 24.5
s2.show() # S-02: 30.1
```

# 3. 속성, 메서드, 생성자

```python
class Counter:
    def __init__(self, start=0):
        self.count = start # 초기 속성 설정
        
    def increase(self, step=1):
        self.count += step # 속성 값 변경
        return self.count

counter = Counter() # __init__(self, 0) 자동 호출
print(counter.increase()) 
print(counter.increase(5))
```

# 4. self, instance

```python
class Sensor:
    def __init__(self, value):
        self.value = value
        
    def add(self, amount):
        self.value += amount

s1 = Sensor(10)
s2 = Sensor(100)
s1.add(5)
print(s1.value, s2.value)
```

# 5. 클래스 변수, 인스턴스 변수

```python
class Sensor:
    count = 0 

    def __init__(self, sensor_id):
        self.sensor_id = sensor_id
        Sensor.count += 1 

s1 = Sensor('S1')
s2 = Sensor('S2')
print(Sensor.count)
print(s1.sensor_id, s2.sensor_id)
```

# 6. 상속

```python
class Sensor:
    def __init__(self, sensor_id):
        self.sensor_id = sensor_id

    def show(self):
        print(f'센서 ID: {self.sensor_id}')

class TemperatureSensor(Sensor):
    def to_fahrenheit(self, celsius):
        return celsius * 9 / 5 + 32

ts = TemperatureSensor('T-01')
ts.show()
print(ts.to_fahrenheit(25))
```

# 7. super()

```python
class Sensor:
    def __init__(self, sensor_id):
        self.sensor_id = sensor_id

class TemperatureSensor(Sensor):
    def __init__(self, sensor_id, unit='C'):
        super().__init__(sensor_id) 
        self.unit = unit 

ts = TemperatureSensor('T-01', unit='C')
print(ts.sensor_id, ts.unit)
```

# 8. Method Overriding

```python
class Sensor:
    def read(self):
        return '측정값 없음'

class TemperatureSensor(Sensor):
    def read(self):
        return '24.5 C'

class HumiditySensor(Sensor):
    def read(self):
        return '60 %'

print(TemperatureSensor().read())
print(HumiditySensor().read())
```

# 9. 다중 상속, Method Round Order

```python
class Logger:
    def info(self):
        print('로그 기록')

class Sensor:
    def read(self):
        print('측정')

class SmartSensor(Logger, Sensor): # 다중 상속
    pass

s = SmartSensor()
s.info() # Logger의 메서드
s.read() # Sensor의 메서드
print(SmartSensor.__mro__) # 탐색 순서 확인
```

# 10. 다형성

```python
class TemperatureSensor:
    def read(self):
        return '24.5 C'

class HumiditySensor:
    def read(self):
        return '60 %'

sensors = [TemperatureSensor(), HumiditySensor()]

for sensor in sensors:
    print(sensor.read()) # 클래스가 달라도 동일하게 호출
```

# 11. 캡슐화

```python
class Sensor:
    def __init__(self, value):
        self.__value = value # private 속성 -> __value

    def get_value(self): # getter 메서드로 접근 제공
        return self.__value

s = Sensor(24.5)
print(s.get_value()) # 24.5
# print(s.__value)
```

# 12. 특수 메서드

```python
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id; self.value = value

    def __str__(self): # print(객체) 시 출력될 문자열
        return f'Sensor[{self.sensor_id}]={self.value}'
    
    def __eq__(self, other): # == 비교 동작 재정의
        return self.value == other.value

s1 = Sensor('S1', 24.5); s2 = Sensor('S2', 24.5)
print(s1) # Sensor[S1]=24.5
print(s1 == s2) # True (값이 같으므로)
```

# 예제

```python
# No 1.
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id
        self.value = value
    
    def show(self):
        print(f'{self.sensor_id}: {self.value}')

    def read(self):
        return "측정값 없음"

s1 = Sensor('S-01', 24.5)
s1.show() # S-01: 24.5
```

```python
# No 2.
class TemperatureSensor(Sensor):
    def __init__(self, sensor_id, value):
        super().__init__(sensor_id, value)

    def to_fahrenheit(self):
        return self.value * 9 / 5 + 32

ts = TemperatureSensor('T-01', 25)
print(ts.sensor_id, ts.to_fahrenheit())
```

```python
# No 3.
class TemperatureSensor(Sensor):
    def __init__(self, sensor_id, value):
        super().__init__(sensor_id, value)

    def to_fahrenheit(self):
        return self.value * 9 / 5 + 32

    def read(self):
        return str(self.value)

class HumiditySensor(Sensor):
    def __init__(self, sensor_id, value):
        super().__init__(sensor_id, value)

    def read(self):
        return str(self.value)

print(TemperatureSensor(1, 24.5).read())
print(HumiditySensor(2, 60).read())
```

```python
# No 5.
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id
        self.__value = value
    
    def show(self):
        print(f'{self.sensor_id}: {self.__value}')

    def read(self):
        return str(self.__value)

    def getValue(self):
        return self.__value

    def __str__(self):
        return f'Sensor[{self.sensor_id}] = {self.__value}'

s = Sensor('S-01', 24.5)
print(s.getValue()) # 24.5
print(s) 
```
