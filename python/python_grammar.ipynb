# 10. 조건문

```python
speed = 87 # 현재 속도를 저장한다.

if speed >= 100: # 100 이상인지 먼저 검사한다.
    level = 'DANGER' # 참이면 위험 상태를 저장한다.
elif speed >= 80: # 첫 조건이 거짓일 때 80 이상인지 검사한다.
    level = 'WARNING' # 참이면 경고 상태를 저장한다.
else: # 앞 조건이 모두 거짓인 경우다.
    level = 'NORMAL' # 정상 상태를 저장한다.

print(level)
```

# 13. for, while

```python
values = [21.1, 22.0, 23.4] 
total = 0

for value in values:
    total += value 

print(total / len(values)) 

n = 1 
sum_to_10 = 0

while n <= 10: 
    sum_to_10 += n; n += 1 

print(sum_to_10)
```

# 14. 함수
 * def total(*args): 값을 튜플로 받음
 * def make(**kwargs): 값을 dict로 받음

```python
def calc_rect(x, y):
    return x * y

rect_x = 10
rext_y = 20

print("사각형 x의 길이: ", rect_x)
print("사각형 y의 길이: ", rext_y)

print("사각형 넓이: ", calc_rect(rect_x, rext_y))
```

```python
def f(x):
    return 2 * x + 7

def g(x):
    return x ** 2

x = 2
print(f(x) + g(x) + f(g(x)) + g(f(x)))
```

```python
def a_rect_area():
    print(5 * 7)

def b_rect_area(x, y):
    print(x * y)

def c_rect_area():
    return (5 * 7)

def d_rect_area(x, y):
    return (x * y)
```

```python
a_rect_area()
b_rect_area(5, 7)
print(c_rect_area())
print(d_rect_area(5, 7))
```

## 가변인수

```python
def asterisk_test(a, b, *args):
    return a + b + sum(args)

print(asterisk_test(1, 2, 3, 4, 5))
```

```python
def asterisk_test(a, b, *args):
    print(args)

print(asterisk_test(1, 2, 3, 4, 5))
```

## 키워드 가변 인수

```python
def kwargs_test(**kwargs):
    print(kwargs)
    print("First value is {first}".format(**kwargs))
    print("Second value is {second}".format(**kwargs))
    print("Third value is {third}".format(**kwargs))

kwargs_test(first = 3, second = 4, third = 5)
```

```python
# 9번 문제
def my_max(arr):
    res = -999999
    for i in arr:
        if (i > res):
            res = i
    return res

def my_min(arr):
    res = 999999
    for i in arr:
        if (i < res):
            res = i
    return res

def my_avg(arr):
    res = 0
    for i in arr:
        res += i
    return res / len(arr)

arr = [1, 2, 3, 4, 5, 6, 9]

print(my_max(arr))
print(my_min(arr))
print(my_avg(arr))
```

```python
# 10번 문제
def temp_state(arr):
    my_dict = {}
    for i in arr:
        if (i >= 38):
            my_dict[i] = "위험"
        elif (i >= 30):
            my_dict[i] = "주의"
        else:
            my_dict[i] = "정상"

    return my_dict

arr = [35, 38, 26, 0, 40]
print(temp_state(arr))
```
