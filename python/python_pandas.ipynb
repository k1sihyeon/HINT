# 1. Pandas

```python
import pandas as pd

# Series: 1차원 라벨링 배열
s = pd.Series([21.5, 30.2, 26.1], index=['S1','S2','S3'])
print(s)

# DataFrame: 2차원 표
df = pd.DataFrame({'sensor':['S1','S2'], 'temp':[21.5,30.2]})
print(df)
```

# 2. Series 생성, 조회

```python
import pandas as pd

temp = pd.Series([21.5, 30.2, 26.1], index=['S1', 'S2', 'S3'])

print(temp['S2']) # 인덱스 라벨로 조회 -> 30.2
print(temp.iloc[0]) # 위치(정수)로 조회 -> 21.5
print(temp.values) # 값만 ndarray로
print(temp.index) # 인덱스(라벨) 목록
```

# 5. 열/행 선택

# 6. 조건 필터링/ 불리언 인덱싱

```python
import pandas as pd

df = pd.DataFrame({'zone':['A','A','B','B'], 'temp':[22,31,25,33]})

warning = df[df['temp'] >= 30] # 단일 조건
print(warning)

# 여러 조건: 각 조건을 괄호로 감싸고 & 로 결합
cond = (df['zone'] == 'B') & (df['temp'] >= 30)
print(df[cond])

# isin(): 목록에 포함된 값만 필터링
print(df[df['zone'].isin(['A'])])
```

# 7. 열 추가/수정/삭제

```python
import pandas as pd

df = pd.DataFrame({'sensor':['S1','S2','S3'], 'temp':[23.0,34.2,41.0]})

def status(t):
    if t >= 40: return '위험'
    if t >= 30: return '주의'
    return '정상'

df['status'] = df['temp'].apply(status) # 새 열 추가
df = df.drop('sensor', axis=1) # 열 삭제(예시)
print(df)
```

# 8. 결측치 NaN 처리

```python
import pandas as pd

df = pd.DataFrame({'temp':[22.0, None, 22.0, 31.5],
    'sensor':['S1','S1','S1','S2']})

print('결측치 수:\n', df.isna().sum())

df['temp'] = df['temp'].fillna(df['temp'].median()) # 중앙값 대체
df = df.drop_duplicates() # 완전히 같은 행 제거
print(df)
```

# 9. 정렬

```python
import pandas as pd

df = pd.DataFrame({'sensor':['S1','S2','S3'], 'temp':[26.1,21.5,30.2]})

asc = df.sort_values(by='temp')                     # 오름차순(기본)
desc = df.sort_values(by='temp', ascending=False)   # 내림차순

print(asc)
print(desc)
```

# 10. 그룹별 집계; group by

```python
import pandas as pd

df = pd.DataFrame({'zone':['A','A','B','B'], 'temp':[22,31,25,33]})

print(df.groupby('zone')['temp'].mean())                # 구역별 평균
print(df.groupby('zone')['temp'].agg(['mean','max']))   # 여러 통계 동시 계산
```

# 11. 시간 데이터 처리; datatime, resample

```python
import pandas as pd

df = pd.DataFrame({
    'time': pd.to_datetime(['2026-08-01 09:00','2026-08-01 09:05','2026-08-01 10:00']),
    'temp': [22, 24, 26]})

hourly = df.set_index('time').resample('1h')['temp'].mean()
print(hourly) # 1시간 단위 평균
```

# 12. 데이터 결합; merge/concat

```python
import pandas as pd

info = pd.DataFrame({'sensor':['S1','S2'], 'location':['1F','2F']})
log = pd.DataFrame({'sensor':['S1','S2'], 'temp':[22.1,30.5]})

merged = pd.merge(log, info, on='sensor') # 공통 키로 결합
print(merged)
```

# 13. 파일 저장

```python
import pandas as pd

df = pd.DataFrame({'sensor_id':['S1','S1','S2'], 'temp':[22,33,35]})

summary = df.groupby('sensor_id', as_index=False)['temp'].agg(['mean','max'])
summary.to_csv('sensor_summary.csv', index=False, encoding='utf-8-sig')
print(summary)
```

# 14. 예제

### 1) CSV 로그 불러오기와 타입 변환

```python

import pandas as pd
from io import StringIO

csv_text = '''
time,sensor_id,temp,humidity
2026-08-01 09:00,S1,22.1,55
2026-08-01 09:05,S1,22.5,56
2026-08-01 09:00,S2,31.2,60
'''

df = pd.read_csv(StringIO(csv_text))
df['time'] = pd.to_datetime(df['time']) # 문자열 -> datetime

print(df.dtypes)
print(df.head())
```

### 2) 결측치·중복 처리

```python
import pandas as pd

df = pd.DataFrame({'temp':[22.0, None, 22.0, 31.5],
    'sensor':['S1','S1','S1','S2']})
print('결측치 수:\n', df.isna().sum())

df['temp'] = df['temp'].fillna(df['temp'].median())
df = df.drop_duplicates()
print(df)
```

### 3) 조건 필터 + apply로 상태 분류

```python

import pandas as pd

df = pd.DataFrame({'sensor_id':['S1','S2','S3'], 'temp':[23.0, 34.2, 41.0]})

def status(temp):
    if temp >= 40:
        return '위험'
    if temp >= 30:
        return '주의'
    return '정상'

df['status'] = df['temp'].apply(status)
print(df)
```

### 4) 시간 단위 집계(resample)

```python

import pandas as pd

df = pd.DataFrame({
'time': pd.to_datetime(['2026-08-01 09:00','2026-08-01 09:05','2026-08-01 10:00']),
'temp': [22, 24, 26]})

hourly = df.set_index('time').resample('1h')['temp'].mean()
print(hourly)
```

### 5) 그룹 집계 결과와 경고 목록 저장

```python

import pandas as pd

df = pd.DataFrame({'sensor_id':['S1','S1','S2'], 'temp':[22,33,35]})

summary = df.groupby('sensor_id', as_index=False)['temp'].agg(['mean','max']).reset_index()
alerts = df[df['temp'] >= 30]
print(alerts)

summary.to_csv('sensor_summary.csv', index=False, encoding='utf-8-sig')
alerts.to_csv('sensor_alerts.csv', index=False, encoding='utf-8-sig')
print(summary)
```

# 15. 확인 문제

```python
# 1) sensor, temp, humidity 열을 가진 DataFrame을 만들고 구조를 info()로 확인
import pandas as pd

df = pd.DataFrame({'sensor': ['S1', 'S2', 'S3'], 'zone': ['A', 'A', 'B'], 'temp':[23.0, 34.2, 41.0], 'humidity':[55, 60, 58]})
print(df.info())

# 2) temp가 30 이상인 행만 필터링하라.
warning = df[df['temp'] >= 30]
print(warning)

# 3) apply()로 온도값에 따른 상태(정상/주의/위험) 열을 추가하라.
def status(t):
    if t >= 40: return '위험'
    if t >= 30: return '주의'
    return '정상'

df['status'] = df['temp'].apply(status)
print(df)

# 4) zone별 temp 평균과 최대값을 groupby+agg로 구하라.
summary = df.groupby('zone')['temp'].agg(['mean', 'max'])
print(summary)

# 5) 결과를 utf-8-sig 인코딩의 CSV로 저장하라.
df.to_csv('results.csv', index=False, encoding='utf-8-sig')
```
