```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. 샘플 데이터프레임 생성
np.random.seed(42)
dates = pd.date_range('2026-01-01', periods=100)
df = pd.DataFrame({
'Sales': np.random.randint(100, 500, size=100).cumsum(),
'Profit': np.random.randint(10, 50, size=100).cumsum()
}, index=dates)

# 2. 데이터 시각화 (JupyterLab 차트 출력)
plt.figure(figsize=(9, 3.8))
plt.plot(df.index, df['Sales'], label='Cumulative Sales', color='#da5b00', linewidth=2)
plt.plot(df.index, df['Profit'], label='Cumulative Profit', color='#2a9d8f', linewidth=2)
plt.title('2026 Business Growth Metrics', fontsize=12, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Amount (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# 1. Sine 데이터 생성
x = np.arange(0, 10, 0.25)
y = np.sin(x)

# 2. 이미지 로드 (디렉토리에 Socrates.png 필요)
img = mpimg.imread('./data/Socrates.png')

# 3. Subplot(1, 2, x) 생성 및 시각화
plt.figure(figsize=(10, 4))

# Left: Sine Graph
plt.subplot(1, 2, 1)     # 빨간색 뭔형 마커와 실선
plt.plot(x, y, 'ro-')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sine Function')
plt.grid(True)

# Right: Image Display
plt.subplot(1, 2, 2)
plt.imshow(img)
plt.title('Socrates.png')
plt.axis('off')            # 축 눈금 제거

plt.tight_layout()
plt.show()
```
